Jupyter supports a few different types of cells, including **Markdown** cells and **Code** cells, as described below.

### Markdown cells

**Markdown** cells include formatted text for documentation. The name "[Markdown](https://www.markdownguide.org/)" is a play on the word "[markup](https://en.wikipedia.org/wiki/Markup_language)", which is a class of computer languages that specify how text and images should appear on a page. Another common markup language is **HTML**, or **H**yper**T**ext **M**arkup **L**anguage, which browsers use to format web pages. Markdown is much simpler than HTML and is designed for basic formatting. It is widely used in technical documentation. 

&#x1F4CC; *Note*: You won't need to learn how to write Markdown syntax in PHYS 132, but you will learn to recognize it and modify it.

---
&#x27A1; Double-click on this cell to view the Markdown syntax that produces the formatted text. Notice that the character sequence `&#x27A1;` represents the [Unicode "Black Right Arrow" symbol](https://www.unicodepedia.com/unicode/dingbats/27a1/black-rightwards-arrow/). Enclosing text by single asterisks (`*`) formats it in *italics*. Enclosing text by double asterisks (`**`) formats it in **boldface**. Enclosing text by triple asterisks (`***`) formats it in ***boldface italics***.
